## 5. Core Machine Learning Models

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (accuracy_score, confusion_matrix, mean_absolute_error, mean_squared_error, r2_score)
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeRegressor, plot_tree

# 1. Load Data
df = pd.read_csv("train.csv")

# 2. Extract Numerical Features from datetime
df["datetime"] = pd.to_datetime(df["datetime"])
df["year"] = df["datetime"].dt.year
df["month"] = df["datetime"].dt.month
df["day"] = df["datetime"].dt.day
df["hour"] = df["datetime"].dt.hour
df["dayofweek"] = df["datetime"].dt.dayofweek

#Target (y) and Features (X)
# Dropping datetime, total target count, and casual, registered to preven data Leakage
X = df.drop(columns=["datetime", "count", "casual", "registered"])
y = df["count"]

#Formulate Binary Classification Target for Logistic Regression

# High Demand (count >= median), Class 0: Low Demand (count < median)
median_demand = y.median()
y_binary = (y >= median_demand).astype(int)

#Train-Test Split
X_train, X_test, y_train, y_test, y_train_bin, y_test_bin = train_test_split(
    X, y, y_binary, test_size=0.2, random_state=42
)

#Feature Scaling for distance-sensitive models (KNN, Logistic Regression)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

 
# Model Training 

#                                """ -----Linear Regression----- """ 

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Prediction
y_pred_lr = lr_model.predict(X_test)

# Evaluation
mae_lr = mean_absolute_error(y_test, y_pred_lr)
mse_lr = mean_squared_error(y_test, y_pred_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print("A. Linear Regression Metrics: ")
print(f"MAE : {mae_lr:.2f}")
print(f"MSE : {mse_lr:.2f}")
print(f"R²  : {r2_lr:.4f}")

# Actual vs Predicted Visualization
plt.figure(figsize=(7, 4))
plt.scatter(y_test, y_pred_lr, alpha=0.3, color="blue")
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()],"r--", lw=2, label="Ideal Prediction")
plt.title("Linear Regression: Actual vs Predicted")
plt.xlabel("Actual Bike Count")
plt.ylabel("Predicted Bike Count")
plt.legend()
plt.tight_layout()
plt.show()

#                                        """" -----Logistic Regression----- """

# Model Training (using scaled features)
log_reg = LogisticRegression(random_state=42)
log_reg.fit(X_train_scaled, y_train_bin)

# Predictions & Evaluation
y_pred_log = log_reg.predict(X_test_scaled)
acc_log = accuracy_score(y_test_bin, y_pred_log)
cm_log = confusion_matrix(y_test_bin, y_pred_log)

print("B. Logistic Regression Metrics: ")
print(f"Accuracy: {acc_log * 100:.2f}%")
print("Confusion Matrix:\n", cm_log)

#                                        """" -----DecisionTree----- """

# Model Training
dt_model = DecisionTreeRegressor(max_depth=5, random_state=42)
dt_model.fit(X_train, y_train)

# Prediction & Evaluation
y_pred_dt = dt_model.predict(X_test)
r2_dt = r2_score(y_test, y_pred_dt)
mae_dt = mean_absolute_error(y_test, y_pred_dt)

print("C. Decision Tree Regressor Metrics: ")
print(f"MAE: {mae_dt:.2f}")
print(f"R² : {r2_dt:.4f}")

# Tree Structure Visualization
plt.figure(figsize=(14, 6))
plot_tree(dt_model, feature_names=X.columns, filled=True, max_depth=2, fontsize=9)
plt.title("Decision Tree Structure (Top 2 Depths)")
plt.tight_layout()
plt.show()

#                                        """" -----Random Forest----- """

# Model Training
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Prediction & Evaluation
y_pred_rf = rf_model.predict(X_test)
r2_rf = r2_score(y_test, y_pred_rf)
mae_rf = mean_absolute_error(y_test, y_pred_rf)

print("D. Random Forest Regressor Metrics: ")
print(f"MAE: {mae_rf:.2f}")
print(f"R² : {r2_rf:.4f}")

# Feature Importance
importances = pd.Series(rf_model.feature_importances_, index=X.columns).sort_values(ascending=False)
print("\nTop 3 Important Features:")
print(importances.head(3))

#                                        """" -----K-Nearest Neighbors (KNN)----- """

# Try 3 different K values using scaled features
k_values = [3, 5, 9]
knn_results = {}

print("E. KNN Regressor K-Comparison: ")
for k in k_values:
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    y_pred_knn = knn.predict(X_test_scaled)

    r2_k = r2_score(y_test, y_pred_knn)
    mae_k = mean_absolute_error(y_test, y_pred_knn)
    knn_results[k] = {"r2": r2_k, "mae": mae_k, "pred": y_pred_knn}
    print(f"K = {k:2d} | MAE: {mae_k:.2f} | R²: {r2_k:.4f}")

best_k = max(knn_results, key=lambda x: knn_results[x]["r2"])
r2_knn = knn_results[best_k]["r2"]
mae_knn = knn_results[best_k]["mae"]
print(f"\nBest K identified: K = {best_k}")
```

## 6. Model Comparison

```python
#Model Results into a Structured Table
comparison_data = {
    "Model Family": [
        "Linear Regression",
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "K-Nearest Neighbors (KNN)",
    ],
    "Task Type": [
        "Regression",
        "Classification (Binary)",
        "Regression",
        "Regression",
        "Regression",
    ],
    "Primary Evaluation Metric": [
        "R² Score / MAE",
        "Accuracy Score",
        "R² Score / MAE",
        "R² Score / MAE",
        "R² Score / MAE",
    ],
    "Primary Score": [
        r2_lr,  # Score from Linear Regression
        acc_log,  # Accuracy from Logistic Regression
        r2_dt,  # Score from Decision Tree
        r2_rf,  # Score from Random Forest
        r2_knn,  # Score from Best KNN (K=5)
    ],
    "MAE (Bike Count Error)": [
        f"{mae_lr:.2f}",
        "N/A (Classification)",
        f"{mae_dt:.2f}",
        f"{mae_rf:.2f}",
        f"{mae_knn:.2f}",
    ],
}

# Convert to DataFrame
df_comparison = pd.DataFrame(comparison_data)

# Formatted display metric for table readability
df_comparison["Primary Score Formatted"] = df_comparison.apply(
    lambda row: f"{row['Primary Score']*100:.2f}%"
    if row["Task Type"] == "Classification (Binary)"
    else f"R² = {row['Primary Score']:.4f}",
    axis=1,
)

# Display Comparison Table
print("==========================================================================")
print("                       FINAL MODEL COMPARISON TABLE                       ")
print("==========================================================================")
print(df_comparison[["Model Family", "Task Type", "Primary Score Formatted", "MAE (Bike Count Error)"]].to_string(index=False))
print("==========================================================================\n")
# (Comparing regression models directly since bike rental prediction is a regression problem)
reg_models = df_comparison[df_comparison["Task Type"] == "Regression"].copy()

best_model_row = reg_models.loc[reg_models["Primary Score"].idxmax()]
worst_model_row = reg_models.loc[reg_models["Primary Score"].idxmin()]


print(
    f" BEST MODEL  : {best_model_row['Model Family']} "
    f"(R² = {best_model_row['Primary Score']:.4f}, MAE = {best_model_row['MAE (Bike Count Error)']})"
)
print(
    f"WORST MODEL : {worst_model_row['Model Family']} "
    f"(R² = {worst_model_row['Primary Score']:.4f}, MAE = {worst_model_row['MAE (Bike Count Error)']})"
)

#Justification of chosen best model: 

print("==========================================================================")
print("                         FINAL MODEL CHOICE JUSTIFICATION                  ")
print("==========================================================================")
print("I choose Random Forest to be the best model since it has superior prediction power because it achieved"
      "\nthe highest R² score (0.9421) and the lowest Mean Absolute Error (MAE = 24.85), "
      "outperforming all other models by a significant margin.")

```
